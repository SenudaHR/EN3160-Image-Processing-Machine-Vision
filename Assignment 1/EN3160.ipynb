```python
#Question 1

import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

f = cv.imread("/content/emma.jpg", cv.IMREAD_GRAYSCALE)

# Check if the image was loaded.
if f is None:
    print("Error: Could not load image")
else:
    t1=np.linspace(0,50,50).astype("uint8")
    t2=np.linspace(100,255,100).astype("uint8")
    t3=np.linspace(150,255,106).astype("uint8")

    tr=np.concatenate((t1,t2,t3)).astype("uint8")

    img=cv.LUT(f,tr)
    cv2_imshow(img)
    cv.waitKey(0)
    cv.destroyAllWindows()
```
